## Aim
Build a document-grounded assistant using RAG. A small local retrieval model is used; no GPU is required.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "FlowGuard is an IoT water monitoring project that detects leaks.",
    "FlowGuard uses an ESP32, water flow sensor, MQTT, and a dashboard.",
    "Trust Index is a framework for checking hallucinations in RAG systems.",
    "AIML means Artificial Intelligence and Machine Learning."
]
vectorizer = TfidfVectorizer()
doc_vectors = vectorizer.fit_transform(documents)

def retrieve(question):
    qv = vectorizer.transform([question])
    scores = cosine_similarity(qv, doc_vectors)[0]
    best = scores.argsort()[-2:][::-1]
    return " ".join(documents[i] for i in best)

question = input("Ask a question about the documents: ")
context = retrieve(question)
print("Retrieved context:", context)
```

```python
# CPU-friendly answer generation with a small pretrained FLAN-T5 model.
from transformers import pipeline
assistant = pipeline("text2text-generation", model="google/flan-t5-small", device=-1)
prompt = "Answer only from this context. If not known, say I do not know. Context: " + context + " Question: " + question
print("Assistant:", assistant(prompt, max_new_tokens=70)[0]["generated_text"])
```
