## Aim
Demonstrate a text prompt to video workflow.

**CPU limitation:** Current open-source text-to-video foundation models are generally too large for practical inference on a CPU-only i7 PC. This notebook creates a short prompt-labelled MP4 slideshow so the practical can be demonstrated locally. It is not neural text-to-video generation. Use the optional model workflow on a GPU machine.

```python
%pip install -q pillow imageio imageio-ffmpeg
```

```python
from PIL import Image, ImageDraw
import imageio.v2 as imageio
from IPython.display import Video, display

prompt = input("Enter a video prompt: ")
frames = []
for i in range(24):
    img = Image.new("RGB", (512, 288), (235, 240, 245))
    draw = ImageDraw.Draw(img)
    x = 30 + i*15
    draw.ellipse((x, 100, x+60, 160), fill=(50, 120, 200))
    draw.text((20, 20), "Prompt: " + prompt[:55], fill=(0,0,0))
    frames.append(img)

imageio.mimsave("cpu_video_demo.mp4", frames, fps=8)
display(Video("cpu_video_demo.mp4", embed=True))
print("Saved cpu_video_demo.mp4")
```
