## Aim
Generate music from text using pretrained MusicGen.

**CPU limitation:** MusicGen is a large foundation model. On an i7 CPU it can be extremely slow and may require substantial RAM. The first cell below creates a simple playable tone sequence as a guaranteed CPU demonstration. The optional MusicGen section is the actual pretrained model.

```python
%pip install -q numpy soundfile
```

```python
import numpy as np
import soundfile as sf
from IPython.display import Audio, display

prompt = input("Describe the music you want: ")
sample_rate = 22050
duration = 4
t = np.linspace(0, duration, int(sample_rate*duration), endpoint=False)
# Simple CPU audio demo: prompt is recorded in the output description;
# this is not neural MusicGen generation.
audio = 0.2*np.sin(2*np.pi*440*t) + 0.1*np.sin(2*np.pi*554.37*t)
sf.write("cpu_music_demo.wav", audio, sample_rate)
display(Audio(audio, rate=sample_rate))
print("Prompt:", prompt)
print("Saved cpu_music_demo.wav")
```
