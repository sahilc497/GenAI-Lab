## Aim
Generate an image from a text prompt.

**CPU limitation:** Stable Diffusion is extremely slow and memory-intensive without a GPU. This notebook first provides a quick CPU-safe prompt-to-image placeholder demonstration using Pillow. The optional pretrained Stable Diffusion cell can be attempted only on a PC with sufficient RAM and patience; it may take a very long time.

```python
%pip install -q pillow matplotlib
```

```python
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

prompt = input("Enter an image prompt: ")
img = Image.new("RGB", (512, 512), "white")
draw = ImageDraw.Draw(img)
draw.rectangle((30, 30, 482, 482), outline="black", width=3)
draw.text((50, 220), "Text prompt:", fill="black")
draw.text((50, 250), prompt[:55], fill="black")
plt.imshow(img)
plt.axis("off")
plt.show()
img.save("prompt_demo.png")
print("Saved prompt_demo.png")
```
