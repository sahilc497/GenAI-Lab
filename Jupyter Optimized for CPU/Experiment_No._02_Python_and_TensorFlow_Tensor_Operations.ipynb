```python
import tensorflow as tf

a, b = 10, 5
print("Addition:", a+b)
print("Subtraction:", a-b)
print("Multiplication:", a*b)
print("Division:", a/b)

x = tf.constant([1, 2, 3])
y = tf.constant([4, 5, 6])
print("Tensor addition:", tf.add(x, y))
print("Tensor multiplication:", tf.multiply(x, y))

matrix = tf.constant([[1, 2], [3, 4]])
print("Transpose:\\n", tf.transpose(matrix))
print("Matrix multiplication:\\n", tf.matmul(matrix, matrix))
print("Reshape:", tf.reshape(tf.range(6), [2, 3]))

v = tf.Variable([10, 20, 30])
v.assign([40, 50, 60])
print("Updated variable:", v.numpy())
```
