```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, MinMaxScaler

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Age": [20, 21, np.nan, 23, 24],
    "Marks": [80, 90, 75, 85, 95],
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
})
print("Original dataset")
display(df)
print("\\nMissing values")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Gender"] = LabelEncoder().fit_transform(df["Gender"])
df[["Age", "Marks"]] = MinMaxScaler().fit_transform(df[["Age", "Marks"]])

print("\\nPreprocessed dataset")
display(df)
df[["Age", "Marks"]].hist()
plt.show()
sns.heatmap(df[["Age", "Marks", "Gender"]].corr(), annot=True)
plt.show()
```

## Result
Dataset was analyzed and preprocessed using Pandas and Scikit-learn.
