## Aim
Generate music from a text prompt using pretrained MusicGen and save/play the audio. GPU recommended.

```python
import torch, soundfile as sf
from transformers import AutoProcessor, MusicgenForConditionalGeneration
from IPython.display import Audio, display
model_id="facebook/musicgen-small"
processor=AutoProcessor.from_pretrained(model_id)
model=MusicgenForConditionalGeneration.from_pretrained(model_id)
device="cuda" if torch.cuda.is_available() else "cpu"
model=model.to(device)
prompt=input("Describe the music: ")
inputs=processor(text=[prompt],padding=True,return_tensors="pt").to(device)
with torch.no_grad():
 audio=model.generate(**inputs,max_new_tokens=256)
wave=audio[0,0].cpu().numpy()
rate=model.config.audio_encoder.sampling_rate
sf.write("generated_music.wav",wave,rate)
display(Audio(wave,rate=rate))
print("Saved generated_music.wav")
```

## Result
MusicGen generated, played, and saved a short music composition.
