## Aim
Perform basic Python arithmetic and TensorFlow tensor operations.

```python
import tensorflow as tf
a,b=10,5
print("Add:",a+b,"Subtract:",a-b,"Multiply:",a*b,"Divide:",a/b)
x=tf.constant([1,2,3,4]); y=tf.constant([5,6,7,8])
print("Tensor add:",tf.add(x,y))
print("Tensor subtract:",tf.subtract(x,y))
print("Tensor multiply:",tf.multiply(x,y))
print("Tensor divide:",tf.divide(x,y))
m=tf.constant([[1,2],[3,4]])
print("Transpose:",tf.transpose(m))
print("Matrix multiplication:",tf.matmul(m,m))
print("Reshape:",tf.reshape(tf.range(6),[2,3]))
v=tf.Variable([10,20,30]); v.assign([40,50,60]); print("Variable:",v)
print("Random:",tf.random.normal([2,3]))
```
