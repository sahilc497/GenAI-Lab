## Aim
Create a simple question-answering text generation system.

```python
from transformers import pipeline
qa = pipeline("text2text-generation", model="google/flan-t5-small")
question = input("Enter your question: ")
answer = qa(question, max_new_tokens=100)
print("Answer:", answer[0]["generated_text"])
```

## Result
FLAN-T5 generated an answer to the entered question.
