## Aim
Explore, analyze, clean, and preprocess a dataset.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
df = pd.DataFrame({"Name":["A","B","C","D","E"],"Age":[20,21,np.nan,23,24],"Marks":[80,90,75,85,95],"Gender":["Male","Female","Male","Female","Male"]})
display(df)
print(df.info())
display(df.describe())
print(df.isnull().sum())
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Gender"] = LabelEncoder().fit_transform(df["Gender"])
df[["Age","Marks"]] = MinMaxScaler().fit_transform(df[["Age","Marks"]])
display(df)
df[["Age","Marks"]].hist(); plt.show()
sns.heatmap(df[["Age","Marks","Gender"]].corr(), annot=True); plt.show()
```

Dataset analysis, missing-value handling, encoding, normalization, and visualization completed.
