## Aim
Generate a short video directly from a text prompt using a pretrained open-source video model. GPU with substantial VRAM recommended.

```python
import torch
from diffusers import DiffusionPipeline
from diffusers.utils import export_to_video
from IPython.display import Video, display
model_id="damo-vilab/text-to-video-ms-1.7b"
pipe=DiffusionPipeline.from_pretrained(model_id,torch_dtype=torch.float16,variant="fp16")
pipe.enable_model_cpu_offload()
pipe.enable_vae_slicing()
```

```python
prompt=input("Enter a video prompt: ")
result=pipe(prompt,num_frames=16,num_inference_steps=25)
frames=result.frames[0]
export_to_video(frames,"generated_video.mp4",fps=8)
display(Video("generated_video.mp4",embed=True))
print("Saved generated_video.mp4")
```

## Result
A short video is generated, previewed, and saved as MP4. The model is large and may exceed free Colab GPU memory; use a high-memory GPU if needed.
